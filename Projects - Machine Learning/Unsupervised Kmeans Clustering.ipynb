## Data Preparation and Model Discovery

Prepare our data for Kmeans Clustering!

### Missing Data


```python
df.isnull().sum()
```

```python
df[df['Agriculture'].isnull()]['Country']
```

```python
# REMOVAL OF TINY ISLANDS
df[df['Agriculture'].isnull()] = df[df['Agriculture'].isnull()].fillna(0)
```

**-: Check**

```python
df.isnull().sum()
```

**-: Notice climate is missing for a few countries**



```python
#filling-missing-values-by-mean-in-each-group
df['Climate'] = df['Climate'].fillna(df.groupby('Region')['Climate'].transform('mean'))
```

**-: Check again on many elements are missing:**

```python
df.isnull().sum()
```

**-: It looks like Literacy percentage is missing**

```python
df[df['Literacy (%)'].isnull()]
```

```python
#filling-missing-values-by-mean-in-each-group
df['Literacy (%)'] = df['Literacy (%)'].fillna(df.groupby('Region')['Literacy (%)'].transform('mean'))
```

**-: Check again on the remaining missing values:**

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

## Data Feature Preparation

```python
X = df.drop("Country",axis=1)
```

```python
X = pd.get_dummies(X)
```

```python
X.head()
```

### Scaling

**-: Due to some measurements being in terms of percentages and other metrics being total counts (population), we should scale this data first.**

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled_X = scaler.fit_transform(X)
```

```python
scaled_X
```

### Creating and Fitting Kmeans Model




```python
from sklearn.cluster import KMeans
```

```python
ssd = []

for k in range(2,30):
    
    model = KMeans(n_clusters=k)
    
    
    model.fit(scaled_X)
    
    #Sum of squared distances of samples to their closest cluster center.
    ssd.append(model.inertia_)
```

```python
plt.plot(range(2,30),ssd,'o--')
plt.xlabel("K Value")
plt.ylabel(" Sum of Squared Distances")
```

```python
pd.Series(ssd).diff().plot(kind='bar')
```

# Conclusion

####  The K-means analysis groups countries into three clear archetypes differentiated by development and structure: a high-income, service-heavy cluster with high literacy and connectivity and low infant mortality; a lower-income, agriculture-reliant cluster with higher infant and crude mortality and higher birth rates; and a mid-tier, mixed cluster spanning diverse regions and intermediate profiles. Careful preparation—region-wise imputation for Climate and Literacy, targeted handling of island anomalies, scaling, and one-hot encoding—enabled meaningful distance-based segmentation, and elbow diagnostics supported an interpretable choice around K=3 that mapped coherently on a choropleth by ISO code. While results are sensitive to K and K-means’ spherical-cluster assumption, the clusters align with known geographic-development patterns and offer actionable peer groupings for policy benchmarking and market strategy, with recommended next steps including silhouette/CH validation, PCA, removal of region dummies as a robustness check, and exploration of alternative clustering (e.g., GMM, hierarchical) to capture non-spherical structure.
